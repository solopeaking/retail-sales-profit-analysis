```python
#importing libraries
import pandas as pd
```

```python
#loading dataset adn copy to variabel so that real dataset cant be harmed 
data = pd.read_csv("../data/raw/superstore.csv")
df = pd.DataFrame(data)

df_copy = df.copy()
df_copy
```

```python
#viewing datasets and checking null values
df_copy
df_copy.isnull().sum()
df.head(10)
df.tail(808)
```

```python
#selecting rows to work on after checking the null values 

data_work = df_copy.iloc[0:9001]
data_work

data_work.isnull().sum()

mean_postal = data_work["Postal Code"].mean()
null_postal = round(mean_postal , 0)

data_work["Postal Code"] = data_work["Postal Code"].fillna(null_postal)

data_work.isnull().sum()


#so there is no null values in this 9000 rows and can be used for further data analysis
```

```python
#checking duplicate records and numerical consistency
data_work.duplicated().sum()             #0
data_work.info()
data_work.describe()
```

```python
#checking columns 

data_work[["Order Date" , "Ship Date"]]
data_work["Order Date"] = pd.to_datetime(data_work["Order Date"])
data_work["Ship Date"] = pd.to_datetime(data_work["Ship Date"])

# data_work.info()

data_work["Order Date"].min()          #2015-01-03
data_work["Order Date"].max()        #2018-12-30
data_work["Ship Date"].min()          #2015-01-07
data_work["Ship Date"].max()           #2019-01-05

#comparing dates of order adn shipping
compare = data_work["Ship Date"] < data_work["Order Date"]     # 0
compare = data_work["Ship Date"] > data_work["Order Date"]     #8534
compare = data_work["Ship Date"] == data_work["Order Date"]    #467
compare.sum() 

#shipping delay in each row
delay = data_work["Ship Date"] - data_work["Order Date"]  
delay.shape
print("Delay analysis : " , " minimum : " , delay.min() , " maximum : " , delay.max() , " average : " , delay.mean() , " median : " , delay.median() )

#percentage of shiping delays
```
